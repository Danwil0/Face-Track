```python
train_df['source_length'] = train_df['Source_Text'].apply(lambda x: len(str(x)) if pd.notna(x) else 0)
train_df['target_length'] = train_df['Target_Text'].apply(lambda x: len(str(x)) if pd.notna(x) else 0)
train_df['source_word_count'] = train_df['Source_Text'].apply(lambda x: len(str(x).split()) if pd.notna(x) else 0)
train_df['target_word_count'] = train_df['Target_Text'].apply(lambda x: len(str(x).split()) if pd.notna(x) else 0)

print("\nSource text length statistics:")
print(train_df.groupby('Language_Code')['source_length'].describe())

print("\nTarget text length statistics:")
print(train_df['target_length'].describe())

print("\nMissing values after feature creation:")
print(train_df.isnull().sum())

def is_placeholder(text):
    if pd.isna(text):
        return True
    unique_chars = set(str(text))
    return len(unique_chars) < 5 and len(str(text)) > 20

train_df['is_source_placeholder'] = train_df['Source_Text'].apply(is_placeholder)
train_df['is_target_placeholder'] = train_df['Target_Text'].apply(is_placeholder)

print(f"Entries with placeholder source text: {train_df['is_source_placeholder'].sum()}")
print(f"Entries with placeholder target text: {train_df['is_target_placeholder'].sum()}")
```

```python
placeholder_by_language = train_df.groupby('Language_Code')['is_source_placeholder'].sum()
print("\nPlaceholder distribution by language:")
print(placeholder_by_language)
print(f"Percentage of placeholders: {placeholder_by_language / language_counts * 100}")

train_clean_df = train_df[~(train_df['is_source_placeholder'] | train_df['is_target_placeholder'])].copy()
print(f"\nOriginal dataset size: {len(train_df)}")
print(f"Clean dataset size: {len(train_clean_df)}")
print(f"Removed {len(train_df) - len(train_clean_df)} rows ({(len(train_df) - len(train_clean_df))/len(train_df)*100:.2f}%)")

clean_language_counts = train_clean_df['Language_Code'].value_counts()
print("\nLanguage distribution in clean dataset:")
print(clean_language_counts)
print(f"Percentage of original data: {clean_language_counts / language_counts * 100}")

plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
sns.boxplot(x='Language_Code', y='source_length', data=train_clean_df)
plt.title('Source Text Character Length by Language')
plt.ylabel('Character Count')

plt.subplot(2, 2, 2)
sns.boxplot(x='Language_Code', y='target_length', data=train_clean_df)
plt.title('Target Text (English) Character Length by Language')
plt.ylabel('Character Count')

plt.subplot(2, 2, 3)
sns.boxplot(x='Language_Code', y='source_word_count', data=train_clean_df)
plt.title('Source Text Word Count by Language')
plt.ylabel('Word Count')

plt.subplot(2, 2, 4)
sns.boxplot(x='Language_Code', y='target_word_count', data=train_clean_df)
plt.title('Target Text (English) Word Count by Language')
plt.ylabel('Word Count')

plt.tight_layout()
plt.show()
```

```python
def preprocess_text(text):
    if pd.isna(text):
        return ""

    text = str(text)

    text = re.sub(r'\s+', ' ', text)

    text = text.strip()

    return text
```

```python
train_clean_df['Source_Cleaned'] = train_clean_df['Source_Text'].apply(preprocess_text)
train_clean_df['Target_Cleaned'] = train_clean_df['Target_Text'].apply(preprocess_text)

print("Sample of preprocessed texts:")
print(train_clean_df[['Source_Cleaned', 'Target_Cleaned', 'Language_Code']].head())
```

```python
from sklearn.model_selection import train_test_split
```

```python
languages = train_clean_df['Language_Code'].unique()
print("Languages in dataset:", languages)

min_language_count = min(clean_language_counts)
balanced_dfs = []

for lang in languages:
    lang_df = train_clean_df[train_clean_df['Language_Code'] == lang]
    if len(lang_df) > min_language_count:
        lang_df = lang_df.sample(min_language_count, random_state=42)
    balanced_dfs.append(lang_df)

balanced_df = pd.concat(balanced_dfs).reset_index(drop=True)
print(f"\nBalanced dataset size: {len(balanced_df)}")
print(balanced_df['Language_Code'].value_counts())

train_data, val_data = train_test_split(balanced_df, test_size=0.1,
                                        stratify=balanced_df['Language_Code'],
                                        random_state=42)

print(f"Training set: {len(train_data)}")
print(f"Validation set: {len(val_data)}")
```

```python
def preprocess_function(examples):
    max_length = 128

    model_inputs = tokenizer(
        examples["Source_Cleaned"],
        max_length=max_length,
        truncation=True,
        padding="max_length",
        return_tensors="pt"
    )

    with tokenizer.as_target_tokenizer():
        labels = tokenizer(
            examples["Target_Cleaned"],
            max_length=max_length,
            truncation=True,
            padding="max_length",
            return_tensors="pt"
        )

    model_inputs["labels"] = labels["input_ids"]
    return model_inputs
```

```python
# Robust helpers (work when tokenizer.lang_code_to_id is missing)
def get_lang_token_id(tokenizer, code: str):
    # Try official attributes
    if hasattr(tokenizer, "lang_code_to_id") and code in tokenizer.lang_code_to_id:
        return tokenizer.lang_code_to_id[code]
    try:
        return tokenizer.get_lang_id(code)
    except Exception:
        pass
    # Search additional special tokens
    extra = tokenizer.special_tokens_map.get("additional_special_tokens", []) or []
    for cand in (code, f"<<{code}>>", f"__{code}__"):
        if cand in extra:
            return tokenizer.convert_tokens_to_ids(cand)
    # Last resort
    return tokenizer.convert_tokens_to_ids(code)

def maybe_prefix_with_lang_token(tokenizer, text: str, code: str):
    extra = tokenizer.special_tokens_map.get("additional_special_tokens", []) or []
    for cand in (code, f"<<{code}>>", f"__{code}__"):
        if cand in extra:
            return f"{cand} {text}"
    return text  # if no explicit lang token available, skip prefix
```

```python
from transformers import DataCollatorForSeq2Seq
data_collator = DataCollatorForSeq2Seq(tokenizer=tokenizer, model=model, padding="longest")

def preprocess_function(batch):
    input_ids, attention_mask, labels = [], [], []
    # Debugging print: Print the type and content of the batch
    # print(f"Batch type: {type(batch)}")
    # print(f"Batch keys: {batch.keys()}")
    # if "Language_Code" in batch:
    #     print(f"Language_Code type in batch: {type(batch['Language_Code'])}")
    #     print(f"Sample Language_Code values in batch: {batch['Language_Code'][:5]}")


    for src, tgt, lang in zip(batch["Source_Cleaned"], batch["Target_Cleaned"], batch["Language_Code"]):
        # Debugging print: Print type and value of lang before processing
        # print(f"Processing - Type of lang: {type(lang)}, Value of lang: {lang}")
        if lang is None:
            # print("Skipping row due to None language code.")
            continue
        src = "" if src is None else str(src)
        tgt = "" if tgt is None else str(tgt)
        try:
            src_code = LANGUAGE_CODES[lang]
        except KeyError:
            # print(f"Skipping row due to unknown language code: {lang}")
            continue

        # Optional: prefix source with its language token if available
        src_text = maybe_prefix_with_lang_token(tokenizer, src, src_code)

        enc = tokenizer(src_text, max_length=128, truncation=True)

        # Set the target language for the tokenizer before the target tokenization
        tokenizer.set_target_languages(['eng_Latn'])

        with tokenizer.as_target_tokenizer():
            dec = tokenizer(tgt, max_length=128, truncation=True)

        input_ids.append(enc["input_ids"])
        attention_mask.append(enc["attention_mask"])
        labels.append(dec["input_ids"])

    return {"input_ids": input_ids, "attention_mask": attention_mask, "labels": labels}
```
